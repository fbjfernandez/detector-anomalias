```python
import pandas as pd

print(f"Pandas versión: {pd.__version__}")

ruta_csv = "../data/raw/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv"
print(f"Ruta configurada: {ruta_csv}")
```

```python
import numpy as np

print("Cargando el dataset... (esto puede tardar unos segundos por el tamaño)")
df = pd.read_csv(ruta_csv)


print(f"Forma original del dataset: {df.shape}")


df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)
print(f"Forma tras limpiar nulos/infinitos: {df.shape}")


display(df.head())


print("\nDistribución de clases (Tráfico Normal vs Ataques):")
print(df[' Label'].value_counts())
```
